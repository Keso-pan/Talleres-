```python
import pandas as pd
import numpy as np
import re


df = pd.read_csv("real_estate_ecuador_200k.csv")


def extraer_tipo(titulo):
    titulo = str(titulo).lower()
    if 'casa' in titulo: return 'Casa'
    if 'departamento' in titulo or 'dpto' in titulo: return 'Departamento'
    if 'suite' in titulo: return 'Suite'
    if 'local comercial' in titulo or 'local' in titulo: return 'Local Comercial'
    return 'Otro'

df['Tipo_Propiedad'] = df['Titulo'].apply(extraer_tipo)


def extraer_ciudad(lugar):
    partes = [p.strip() for p in str(lugar).split(',')]
    if len(partes) >= 3:
        ciudad = partes[-2]
        return re.sub(r'\s*\d+$', '', ciudad)
    return "Desconocido"

df['Ciudad'] = df['Lugar'].apply(extraer_ciudad)

print("Datos cargados y preprocesados correctamente. Total de filas:", len(df))
```

```python

print(" A. PERSPECTIVA FINANCIERA \n")

precio_promedio_general = df['Precio'].mean()
print(f"Precio Promedio General: ${precio_promedio_general:,.2f}\n")

precio_promedio_ciudad = df.groupby('Ciudad')['Precio'].mean()
print("Precio Promedio por Ciudad (Primeras 5):")
print(precio_promedio_ciudad.head(), "\n")


costo_promedio_m2 = (df['Precio'] / df['Area']).mean()
print(f"Costo Promedio por Metro Cuadrado: ${costo_promedio_m2:,.2f}\n")


rango_precios_provincia = df.groupby('Provincia')['Precio'].agg(['min', 'max', 'mean'])
print("Rango de Precios por Provincia (Primeras 5):")
print(rango_precios_provincia.head())
```

```python

print(" B. PERSPECTIVA DE PRODUCTO \n")


distribucion_tipo = df['Tipo_Propiedad'].value_counts()
print("Distribución por Tipo de Propiedad:")
print(distribucion_tipo, "\n")


tamano_promedio_provincia = df.groupby('Provincia')['Area'].mean()
print("Tamaño Promedio (m2) por Provincia (Primeras 5):")
print(tamano_promedio_provincia.head(), "\n")

precio_por_hab = df.groupby('Num. dormitorios')['Precio'].mean().sort_index()
premium_hab = precio_por_hab.diff().dropna()
print("Premium por Habitación Adicional (Diferencia en $ con la habitación anterior):")
print(premium_hab.head())
```

```python

print(" C. PERSPECTIVA GEOGRÁFICA \n")


top_5_caras = df.groupby('Ciudad')['Precio'].mean().sort_values(ascending=False).head(5)
print("Top 5 Ciudades Más Caras:")
print(top_5_caras, "\n")


concentracion_oferta = df['Provincia'].value_counts()
print("Concentración de la Oferta por Provincia (Primeras 5):")
print(concentracion_oferta.head())
```

```python

print("D. SEGMENTACIÓN DE MERCADO \n")


q1 = df['Precio'].quantile(0.25)
q3 = df['Precio'].quantile(0.75)

condiciones = [
    (df['Precio'] < q1),
    (df['Precio'] > q3)
]
df['Tipo de Inmueble'] = np.select(condiciones, ['Económico', 'Lujo'], default='Medio')

print(f"Límites de segmentación -> Económico: < ${q1:,.2f} | Lujo: > ${q3:,.2f}\n")
print("Distribución de Propiedades por Segmento de Mercado:")
print(df['Tipo de Inmueble'].value_counts())
```

```python
import matplotlib.pyplot as plt


print("--- E. DASHBOARD ---\n")


total_propiedades = len(df)


print(f"Total de Propiedades registradas: {total_propiedades:,}")


fig, ax = plt.subplots(figsize=(5, 2))
fig.patch.set_facecolor('#f8f9fa')


ax.text(0.5, 0.7, 'Total de Propiedades', fontsize=14, ha='center', va='center', color='#6c757d', fontweight='bold')
ax.text(0.5, 0.3, f'{total_propiedades:,}', fontsize=28, ha='center', va='center', color='#212529', fontweight='bold')


ax.axis('off')

plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt


df_ciudades_validas = df[df['Ciudad'] != 'Desconocido']


promedios_ciudades = df_ciudades_validas.groupby('Ciudad')['Precio'].mean().sort_values(ascending=False)
promedio_general = df['Precio'].mean()


etiquetas = ['PROMEDIO GENERAL'] + list(promedios_ciudades.index)
valores = [promedio_general] + list(promedios_ciudades.values)


plt.figure(figsize=(20, 8))


barras = plt.bar(etiquetas, valores, color='#4C72B0', edgecolor='black', alpha=0.8)


barras[0].set_color('#C44E52')


plt.title('Precio Promedio de Alquiler: General y por Ciudad', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Ciudad', fontsize=14)
plt.ylabel('Precio Promedio ($)', fontsize=14)


plt.xticks(rotation=90, ha='center', fontsize=10)


for barra in barras:
    altura = barra.get_height()
    plt.text(
        barra.get_x() + barra.get_width() / 2,
        altura + (altura * 0.02),
        f'${altura:,.0f}',
        ha='center',
        va='bottom',
        fontweight='bold',
        fontsize=8,
        rotation=45
    )

plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import StrMethodFormatter


df_limpio = df.dropna(subset=['Area', 'Precio'])
x = df_limpio['Area']
y = df_limpio['Precio']


m, b = np.polyfit(x, y, 1)


plt.figure(figsize=(10, 6), facecolor='white')


plt.scatter(x, y, color='#005587', s=15, alpha=0.4)


plt.plot(x, m*x + b, color='#FF8C00', linewidth=2.5)


estilo_fuente = {'family': 'serif', 'color': '#008000', 'style': 'italic', 'size': 18}

plt.xlabel('Área m²', fontdict=estilo_fuente)
plt.ylabel('Precio', fontdict=estilo_fuente)


plt.gca().yaxis.set_major_formatter(StrMethodFormatter('${x:,.0f}'))


plt.gca().tick_params(axis='y', colors='#008000', labelsize=12)
plt.gca().tick_params(axis='x', colors='#008000', labelsize=12)


plt.grid(True, linestyle='-', color='gray', alpha=0.6)
for spine in plt.gca().spines.values():
    spine.set_color('gray')

plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np


precio_por_hab = df.groupby('Num. dormitorios')['Precio'].mean().sort_index()


premium_hab = precio_por_hab.diff().dropna()


etiquetas = [f"{int(i-1)} a {int(i)}" for i in premium_hab.index]
valores = premium_hab.values


plt.figure(figsize=(11, 6), facecolor='white')


colores = ['#55A868' if v >= 0 else '#C44E52' for v in valores]


barras = plt.bar(etiquetas, valores, color=colores, edgecolor='black', alpha=0.8)


plt.title('Premium por Habitación Adicional', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Transición de Habitaciones', fontsize=12)
plt.ylabel('Diferencia de Precio Promedio ($)', fontsize=12)


plt.axhline(0, color='black', linewidth=1.2)


rango_y = plt.gca().get_ylim()[1] - plt.gca().get_ylim()[0]
margen = rango_y * 0.015

for barra, valor in zip(barras, valores):
    altura = barra.get_height()


    y_pos = altura + margen if valor >= 0 else altura - margen
    va = 'bottom' if valor >= 0 else 'top'

    plt.text(
        barra.get_x() + barra.get_width() / 2,
        y_pos,
        f'${valor:,.0f}',
        ha='center',
        va=va,
        fontweight='bold',
        fontsize=10
    )

plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
